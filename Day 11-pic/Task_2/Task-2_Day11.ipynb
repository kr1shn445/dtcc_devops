```python
Division 1 — Load and inspect
What: Load the CSV and inspect its dimensions, sample rows, columns, and data types.
Why: Understand the structure before changing anything.
Outcome: A baseline view of the original data and its quality issues.
```

```python
import pandas as pd
import numpy as np

# Keep the CSV in the same folder as this notebook.
df_original = pd.read_csv("mall_ecommerce_messy_dataset.csv")
df = df_original.copy()

print("Rows and columns:", df.shape)
display(df.head(10))
df.info()
display(df.describe(include="all").T)
```

```python
Division 2 — Identify and handle missing values
What: Use isnull(), isna(), sum(), dropna(), and fillna() to investigate missing data.
Why: Missing data can break calculations or bias summaries; dropping every incomplete row can discard useful records.
Outcome: A missing-value report and an informed strategy for handling gaps.
```

```python

display(df.isnull().head())
display(df.isna().head())

missing_counts = df.isna().sum().sort_values(ascending=False)
display(missing_counts[missing_counts > 0].to_frame("Missing_Count"))

print("Original shape:", df.shape)
dropna_demo = df.dropna()
print("Shape after dropping rows with any missing value:", dropna_demo.shape)

# We will not drop every incomplete row from the working dataset automatically.
```

```python
Division 3 — Basic statistical analysis:
What: Calculate sums, means, medians, and descriptive statistics before cleaning.
Why: Invalid values and outliers can distort business metrics.
Outcome: Baseline statistics to compare with the cleaned dataset.
```

```python
numeric_cols = ["Quantity", "Unit_Price_INR", "Discount", "Customer_Age", "Customer_Rating"]

print("BEFORE CLEANING")
display(df_original[numeric_cols].agg(["sum", "mean", "median"]).T)
display(df_original[numeric_cols].describe().T)
```

```python
Division 4 — Standardize text and dates
What: Trim spaces, standardize capitalization and known variants, and parse dates.
Why: "chennai", " CHENNAI ", and "Bangalore" should not accidentally be counted as separate categories.
Outcome: Consistent labels and a datetime column; invalid dates become NaT for review.
```

```python
text_cols = ["Mall", "City", "Product", "Category", "Sales_Channel"]

for col in text_cols:
    df[col] = df[col].astype("string").str.strip().str.title()

# Correct known variants using explicit mapping rules.
df["City"] = df["City"].replace({"Bangalore": "Bengaluru", "Hydrabad": "Hyderabad"})
df["Product"] = df["Product"].replace({"Usb C Cable": "USB-C Cable"})
df["Sales_Channel"] = df["Sales_Channel"].replace({"Click And Collect": "Click & Collect"})

# Parse mixed date formats; invalid/impossible dates become NaT.
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="mixed", errors="coerce")

display(df[["Mall", "City", "Product", "Category", "Sales_Channel", "Order_Date"]].head(15))
print("Missing/unparseable dates:", df["Order_Date"].isna().sum())
```

```python
Division 5 — Convert data types and validate numeric ranges
What: Convert numeric columns with pd.to_numeric(), inspect dtypes/info(), and flag impossible values.
Why: Numeric-looking text cannot be safely calculated, and values such as negative quantities or discounts above 100% violate this exercise's business rules.
Outcome: Consistent numeric types and invalid values isolated for review.
```

```python
print("Data types before conversion:")
display(df.dtypes.to_frame("dtype"))
df.info()

numeric_cols = ["Quantity", "Unit_Price_INR", "Discount", "Customer_Age", "Customer_Rating"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# These are example business rules; confirm rules with a data owner in real work.
df.loc[df["Quantity"] <= 0, "Quantity"] = np.nan
df.loc[df["Unit_Price_INR"] <= 0, "Unit_Price_INR"] = np.nan
df.loc[~df["Discount"].between(0, 1), "Discount"] = np.nan
df.loc[~df["Customer_Age"].between(18, 100), "Customer_Age"] = np.nan
df.loc[~df["Customer_Rating"].between(1, 5), "Customer_Rating"] = np.nan

print("Data types after conversion and validation:")
display(df.dtypes.to_frame("dtype"))
display(df[numeric_cols].describe().T)
```

```python
Division 6 — Detect duplicates
What: Count exact duplicate rows and inspect repeated order IDs.
Why: Duplicates can inflate sales, but repeated IDs with conflicting data may require investigation.
Outcome: Exact duplicates are removed; repeated identifiers are flagged rather than blindly discarded.
```

```python
print("Exact duplicate rows:", df.duplicated().sum())

duplicate_ids = df[df["Order_ID"].duplicated(keep=False)].sort_values("Order_ID")
display(duplicate_ids)

df = df.drop_duplicates().copy()
print("Shape after removing exact duplicate rows:", df.shape)

repeated_id_mask = df["Order_ID"].duplicated(keep=False)
print("Rows with repeated Order_IDs needing review:", repeated_id_mask.sum())
display(df.loc[repeated_id_mask].sort_values("Order_ID"))
```

```python
Division 7 — Detect and handle outliers
What: Use describe() and the IQR method to flag potential outliers in quantity and unit price.
Why: Extreme values can distort averages and revenue, but an unusual value is not automatically wrong.
Outcome: Potential outliers are displayed for investigation and an explicit handling decision.
```

```python
outlier_cols = ["Quantity", "Unit_Price_INR"]
display(df[outlier_cols].describe().T)

def iqr_outlier_report(data, column):
    q1 = data[column].quantile(0.25)
    q3 = data[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    mask = (data[column] < lower) | (data[column] > upper)
    print(f"{column}: Q1={q1:.2f}, Q3={q3:.2f}, IQR={iqr:.2f}")
    print(f"Lower bound={lower:.2f}, upper bound={upper:.2f}, outliers={mask.sum()}")
    return data.loc[mask, ["Order_ID", "Product", column]]

for col in outlier_cols:
    display(iqr_outlier_report(df, col))
```

```python
Handling decision: Investigate flagged records first. For this synthetic dataset, ₹99,999 is a suspicious unit price; mark it missing pending verification. Keep other flagged observations unless there is evidence they are errors.
```

```python
# Demonstration only: correct a known suspicious synthetic value pending verification.
df.loc[df["Unit_Price_INR"] == 99999.00, "Unit_Price_INR"] = np.nan

# Optional capping demonstration: create a separate capped quantity column; don't overwrite Quantity.
q1 = df["Quantity"].quantile(0.25)
q3 = df["Quantity"].quantile(0.75)
iqr = q3 - q1
lower_bound = max(1, q1 - 1.5 * iqr)
upper_bound = q3 + 1.5 * iqr
df["Quantity_Capped_Demo"] = df["Quantity"].clip(lower=lower_bound, upper=upper_bound)
display(df[["Order_ID", "Quantity", "Quantity_Capped_Demo"]].head(10))
```

```python
Division 8 — Handle missing values and create Revenue
What: Impute remaining numeric gaps with medians and categorical gaps with modes, then calculate net revenue.
Why: This produces a usable analysis dataset while avoiding revenue calculations on invalid inputs.
Outcome: A cleaned working dataset with a Revenue column.
```

```python
numeric_fill_cols = ["Quantity", "Unit_Price_INR", "Discount", "Customer_Age", "Customer_Rating"]
for col in numeric_fill_cols:
    df[col] = df[col].fillna(df[col].median())

categorical_fill_cols = ["Mall", "City", "Product", "Category", "Sales_Channel"]
for col in categorical_fill_cols:
    modes = df[col].mode(dropna=True)
    if not modes.empty:
        df[col] = df[col].fillna(modes.iloc[0])

# Keep missing dates for review rather than inventing dates.
df["Revenue"] = (df["Quantity"] * df["Unit_Price_INR"] * (1 - df["Discount"])).round(2)

display(df.head(10))
display(df.isna().sum().to_frame("Remaining_Missing_Count"))
print("Revenue formula: Quantity × Unit_Price_INR × (1 − Discount)")
```

```python
Division 9 — Compare before and after cleaning
What: Compare row counts, missing values, data types, and statistical measures.
Why: We need to measure how the cleaning process changed the dataset.
Outcome: A transparent summary of the impact of cleaning.
```

```python
print("Original shape:", df_original.shape)
print("Cleaned shape:", df.shape)

comparison_missing = pd.DataFrame({
    "Before": df_original.isna().sum(),
    "After": df.isna().sum()
}).fillna(0).astype(int)
display(comparison_missing)

print("Original data types:")
display(df_original.dtypes.to_frame("Original dtype"))
print("Cleaned data types:")
display(df.dtypes.to_frame("Cleaned dtype"))

comparison_stats = pd.DataFrame({
    "Before_Mean": df_original[numeric_cols].mean(numeric_only=True),
    "After_Mean": df[numeric_cols].mean(numeric_only=True),
    "Before_Median": df_original[numeric_cols].median(numeric_only=True),
    "After_Median": df[numeric_cols].median(numeric_only=True),
})
display(comparison_stats)
print("Total revenue in cleaned dataset:", round(df["Revenue"].sum(), 2))
```
